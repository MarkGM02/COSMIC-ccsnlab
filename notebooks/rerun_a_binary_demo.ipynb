```python
import pandas as pd
from ccsnlab.data_loading.process_raw import create_sn_info
from ccsnlab.data_loading.rerun_binary import get_init_conds_and_bse_dict, evolve_and_save
```

```python
#read in sample data
sample_data = pd.read_hdf('sample_data.h5', key='data')
```

```python
#using the data loading utilities grab the system initcond and bsedict
met_cosmic = 0.020243
sigma = 265.0
alpha1 = 1.0
qcflag = 5
full_init, bse_dict = get_init_conds_and_bse_dict(sample_data, met_cosmic, sigma, alpha1, qcflag)

#evolve and save the systems -- wraps cosmic's evolve functions and saves the bcm properly
init_cond = full_init[0:5] #just do the first 5 systems for demo
bpp, bcm, _, _ = evolve_and_save(init_cond, bse_dict)

#You now can views the original bcm and bpp as they were!
print(bpp.head(), '\n', bcm.head())
```

```python
#create a sn_info dataframe from the bpp and bcm -- procovide the required fields to make the data properly formatted
sample_mass = sample_data['sample_mass'].min()
singles_mass = sample_data['singles_mass'].min()
sn_info = create_sn_info(bpp, bcm, metallicity=met_cosmic, sample_mass=sample_mass, singles_mass=singles_mass, sigma=sigma, alpha1=alpha1)

#Now do whatever analysis you want on sn_info!
print(sn_info.head())
```
